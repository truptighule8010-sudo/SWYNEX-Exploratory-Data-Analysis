```python
import pandas as pd

df = pd.read_excel("/content/SWYNEX_TECHNOLOGY_Preparation.xlsx")

df
```

```python
# Number of rows and columns
print("Rows and Columns:", df.shape)

# Column names
print("\nColumn Names:")
print(df.columns)

# Data types
print("\nData Types:")
print(df.dtypes)

# Missing values
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
# Statistical Summary

print("Total Orders:", df["Order_ID"].nunique())
print("Total Quantity:", df["Quantity"].sum())
print("Total Sales:", df["Sales"].sum())
print("Total Profit:", df["Profit"].sum())
print("Average Sales:", df["Sales"].mean())
print("Average Profit:", df["Profit"].mean())
print("Minimum Sales:", df["Sales"].min())
print("Maximum Sales:", df["Sales"].max())

# Profit Margin
profit_margin = (df["Profit"].sum() / df["Sales"].sum()) * 100
print("Overall Profit Margin:", round(profit_margin, 2), "%")
```

```python
# Category-wise Sales and Profit

category_analysis = df.groupby("Category")[["Sales", "Profit"]].sum()

print(category_analysis)

# Chart
category_analysis.plot(kind="bar", figsize=(8, 5))

plt.title("Sales and Profit by Category")
plt.xlabel("Category")
plt.ylabel("Amount")
plt.xticks(rotation=0)
plt.legend()
plt.show()
```

```python
# Region-wise Sales and Profit

region_analysis = df.groupby("Region")[["Sales", "Profit"]].sum()

print(region_analysis)

# Chart
region_analysis.plot(kind="bar", figsize=(8, 5))

plt.title("Sales and Profit by Region")
plt.xlabel("Region")
plt.ylabel("Amount")
plt.xticks(rotation=0)
plt.legend()
plt.show()
```

```python
# Product-wise Sales and Profit

product_analysis = df.groupby("Product")[["Sales", "Profit"]].sum().sort_values(
    by="Sales", ascending=False
)

print(product_analysis)

# Chart
product_analysis.plot(kind="bar", figsize=(10, 5))

plt.title("Sales and Profit by Product")
plt.xlabel("Product")
plt.ylabel("Amount")
plt.xticks(rotation=45)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Monthly Sales Trend

df["Order_Date"] = pd.to_datetime(df["Order_Date"])

monthly_sales = df.groupby(
    df["Order_Date"].dt.to_period("M")
)["Sales"].sum()

print(monthly_sales)

# Line Chart
monthly_sales.plot(kind="line", marker="o", figsize=(10, 5))

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.grid()
plt.show()
```

```python
# Quantity Analysis

quantity_analysis = df.groupby("Product")["Quantity"].sum().sort_values(ascending=False)

print("Quantity Sold by Product:")
print(quantity_analysis)

# Chart
quantity_analysis.plot(kind="bar", figsize=(10, 5))

plt.title("Quantity Sold by Product")
plt.xlabel("Product")
plt.ylabel("Total Quantity")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
